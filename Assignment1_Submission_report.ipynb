src/daneel/detection/transit_HIP67522c.py: follows the batman tutorial using the parameters of HIP 67522 c (orbital period, rp/rs, a/rs, inclination, eccentricity, limb darkening). The parameters were found in Barber et al. 2024 as the NASA Exoplanet archive did not have all the up to date information about our selected exoplanet. The file transit_HIP67522c.py computes the transit model and plots it.
